```python
import  numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
t=sns.load_dataset('titanic')
t
```

```python
#HEATMAP

tc = t.corr(numeric_only= True)
tc
```

```python
tc = t.corr(numeric_only= True)
sns.heatmap(tc, cmap= 'coolwarm',annot = True)
plt.show()
```

```python
#DISPLOT
### Distribution plot


sns.displot(t.age, bins = 8)
plt.show()

```

```python
### HISTPLOT


sns.histplot(t.age, bins = 8)
plt.show()
```

```python
### KDE(KERNEL DENSITY ESTIMATION)
# It is a non-parametric method for estimating the probability density function of a random variable, without assuming the specific distribution.

sns.kdeplot(t.age)
plt.show()

```

```python
sns.displot(t.age, bins = 8, kde= True)
plt.show()
```

```python
## BARPLOT
# used to showcase information of a single sns.categorical column



sns.barplot(x= t.sex, y= t.age)
plt.show()
```

```python
sns.barplot(x= t.sex, y= t.age, estimator = 'median')
```

```python
sns.barplot(x= t.sex, y= t.age, hue= t.pclass)
plt.show()
# hue basically further categorizes by some sns.categorical data
```

```python
#### COUNTPLOT
# It is used to show the freqency of a categorical variable

sns.countplot(t.sex)
plt.show()
```

```python
sns.countplot(x = t.pclass, hue = t.embarked)
```

```python
#### SCATTER PLOT
# It is used to show the relationship between two continuous variables.

sns.scatterplot(x= t.age, y= t.fare)
plt.show()
```

```python
sns.scatterplot(x= t.age, y= t.fare, hue= t.sex)
plt.show()
```

```python
sns.scatterplot(x= t.age, y= t.fare, hue= t.sex, style= t.pclass)
plt.show()
```

```python
#### HEXPLOT
# It is used to show the relationship between two continuous variable.
#it differs from scatterplot by how it represents the datapoints concentration

sns.jointplot(x= t.age, y= t.fare)
plt.show()
```

```python

#### JOINtPLOT
# It is combination of scatterplot(inner) and displot(outer)
# It is used to show the relationship between two continuous variable.
# Inner plot options: hex,reg, hist


sns.jointplot(x= t.age, y= t.fare, kind= 'hex')
plt.show()
```

```python
#### REGRESSION PLOT
# Used to represent the relationship betwwen two continuous variables.
# shows how the values are in/de-creasing order along


sns.regplot(x= t.age, y= t.fare)
plt.show()
```

```python
sns.lineplot(x= t.age, y= t.fare)
plt.show()
```

```python
### BOXPLOT

#Used to show the distribution of quantitative data
#Can be used for a sigle continuous column
#Shows relationship between a continuous and a categorical variable



sns.boxplot(x= t.age)
plt.show()
```

```python
sns.boxplot(x= t.age, y= t.sex)
plt.show()
```

```python
sns.boxplot(x= t.age, y= t.sex, hue= t.pclass)
plt.show()
```
