```python
x = np.arange(0, 2*np.pi, 0.5)
y = np.sin(x)
z = np.cos(x)


fig = plt.figure(figsize = (6,2))
ax = fig.add_axes([0,0,1,1])
ax.plot(x, y, color = 'red', marker = 'o', markersize = 5, linestyle = '--')
ax.plot(x, z, color = 'green', marker = '*', markersize = 5, linestyle = '-')
ax.set_xlabel('Angle')
ax.set_ylabel('Sine of Angle')
ax.set_title('Sine Function')
```

```python
x = np.arange(0, 2*np.pi, 0.5)
y = (np.tan(x) +7)
fig = plt.figure(figsize = (6,3))
ax = fig.add_axes([0,0,1,1])
ax.plot(x,y , color='blue', marker = '^', markersize = 5, linestyle = '-')
ax.set_xlabel('Angle')
ax.set_ylabel('Tangent of Angle')
ax.set_title('Tangent Function')
```

```python
x = np.arange(0, 2*np.pi, 0.5)
y = np.log10(x)
z = np.exp(x)
fig = plt.figure(figsize = (6,3))
ax = fig.add_axes([0,0,1,1])
ax.plot(x,y , color='blue', marker = '^', markersize = 5, linestyle = '-')
ax.plot(x,z , color='red', marker = 'o', markersize = 5, linestyle = '--')
ax.set_xlabel('')
ax.set_ylabel('')
ax.set_title('Log/Exponent')
```

```python
x = np.arange(0, 2*np.pi, 0.5)
y = np.sin(x)
z = np.cos(x)


fig = plt.figure(figsize = (6,2))
ax = fig.add_axes([0,0,1,1])
ax.plot(x, y, color = 'blue', marker = 'o', markersize = 5, linestyle = '--', label = 'Sine')
ax.plot(x, np.tan(x), color = 'red', marker = 'o', markersize = 5, linestyle = '--', label= 'Tan')
ax.legend()
```
